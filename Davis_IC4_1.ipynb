```python
%%sql
/* 1. Create the 'olympics' table with STRICT types */
CREATE TABLE olympics (
    event_id INTEGER PRIMARY KEY,
    sport TEXT NOT NULL,
    event_name TEXT NOT NULL,
    year INTEGER NOT NULL,
    medal_winner TEXT,
    country TEXT
) STRICT;

/* 2. Insert at least 3 items into the 'olympics' table */
INSERT INTO olympics (sport, event_name, year, medal_winner, country) VALUES
('Athletics', 'Men''s 100m', 2020, 'Lamont Marcell Jacobs', 'Italy'),
('Swimming', 'Women''s 100m Butterfly', 2020, 'Maggie Mac Neil', 'Canada'),
('Gymnastics', 'Women''s All-Around', 2020, 'Sunisa Lee', 'United States'),
('Cycling', 'Men''s Road Race', 2020, 'Richard Carapaz', 'Ecuador');

/* 3. Select all data from the 'olympics' table to display */
SELECT * FROM olympics;
```

```python
%%sql
/* Filter the olympics table to show events where the country is 'Italy' */
SELECT * FROM olympics WHERE country = 'Italy';
```

```python
%%sql

/* Update the medal winner for event_id 1 (Men's 100m) */
UPDATE olympics SET medal_winner = 'Gianmarco Tamberi' WHERE event_id = 1;

/* Delete the event with event_id 4 (Cycling) */
DELETE FROM olympics WHERE event_id = 4;

/* Display the updated table to confirm changes */
SELECT * FROM olympics;
```
